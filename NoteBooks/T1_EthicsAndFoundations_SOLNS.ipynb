<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  A hospital deploys a machine learning model that outputs a risk score for patient deterioration.
Clinicians are told that the model is “highly accurate”, but they are not given information about which variables influence the prediction or how the score is calculated.
</p>

<h4 style="color:#005EB8; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  📝 Task
</h4>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
Answer the following in 4–5 sentences:

1. Why might this system be described as a black box?
2. What does it mean to trust this system in a clinical context?
3. Is predictive accuracy alone sufficient to justify its use? Why or why not?

✍️ Focus on reasoning rather than technical detail.
</p>

This system can be described as a black box because clinicians are unable to see how the input variables are combined to produce the risk score or which factors are most influential. Trust in a clinical context would require confidence not only that the model is accurate, but that its behaviour is stable, justifiable, and appropriate for patient care. Predictive accuracy alone is not sufficient, because a highly accurate model may still rely on spurious correlations or unfair proxies. Without understanding why a prediction is made, clinicians cannot safely act on it or explain decisions to patients.

<h3 style="color:#041E42; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  🎯 Exercise 2 — Python: Inspecting Data for Hidden Bias
</h3>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  This exercise uses very basic Python to reinforce how bias can enter through data rather than code. You are given a small dataset representing patients and whether they received follow-up care.
</p>

<h4 style="color:#005EB8; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  📝 Task
</h4>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">

1. Count how many male and female patients received follow-up care.

2. Print the result in a clear, readable way.
</p>

```python
patients = [
    {"age": 72, "sex": "F", "follow_up": True},
    {"age": 68, "sex": "M", "follow_up": True},
    {"age": 75, "sex": "F", "follow_up": False},
    {"age": 70, "sex": "M", "follow_up": True},
    {"age": 80, "sex": "F", "follow_up": False},
]
```

```python
# Make a DataFrame and count the number of patients in each category.
df = pd.DataFrame(patients)
# Group by 
category_counts = df.groupby(["sex", "follow_up"]).size().reset_index(name="count")
```

```python
counts = {"F": 0, "M": 0}

for patient in patients:
    if patient["follow_up"]:
        counts[patient["sex"]] += 1

print("Follow-up received:")
print("Female:", counts["F"])
print("Male:", counts["M"])
```

What does this simple example show about how unequal outcomes can arise before any model is trained?

This example shows that unequal outcomes can arise directly from how data is distributed, even before any machine learning model is trained. If one group receives follow-up care less often in the historical data, a model trained on this data may learn and reinforce that pattern. Bias can therefore be embedded in the data collection process itself, not just in modelling choices.

<h3 style="color:#041E42; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  🎯 Exercise 3 — Written: Bias or Confounding?
</h3>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  For each scenario below, state whether the primary issue is algorithmic bias, algorithmic confounding, or both, and briefly explain why.
</p>

<h4 style="color:#005EB8; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  📝 Scenario A
</h4>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  A model trained to predict disease risk performs well overall, but consistently under-predicts risk for one demographic group.
</p>

<h4 style="color:#005EB8; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  📝 Scenario B
</h4>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  A model finds that patients who live closer to a hospital have lower mortality. Further investigation shows these patients are admitted earlier and receive faster treatment.
</p>

Answer A: Algorithmic bias

The model systematically under-predicts risk for one demographic group, indicating unequal performance across subgroups. This suggests a fairness issue arising from data imbalance, feature selection, or model objectives rather than a causal structure.

Answer B: Algorithmic confounding

The apparent relationship between distance from hospital and mortality is driven by an unobserved variable—earlier admission and faster treatment. The model captures a correlation that is not causal, illustrating confounding rather than bias.

<h3 style="color:#041E42; font-family:Source Sans Pro, Verdana, Arial, sans-serif;">
  🎯 Exercise 4 — Python: Plausible ≠ Correct
</h3>

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">
  This exercise mirrors a key risk of generative AI: producing outputs that look reasonable but may be misleading. Consider the following function, which generates a short “clinical summary”:
</p>

```python
def generate_summary(age, condition):
    return f"The patient is {age} years old and presents with {condition}. Prognosis is good."
```

<p style="font-family:Source Sans Pro, Verdana, Arial, sans-serif; color:#333;">

1. Call the function for at least two different inputs.

2. Print the outputs
</p>

```python
print(generate_summary(65, "chest pain"))
print(generate_summary(82, "shortness of breath"))
```

Reflection (3–4 sentences)

- Why might this output appear trustworthy?

- What important clinical information is missing?

- How does this relate to concerns about large language models in healthcare?

The output appears trustworthy because it uses confident clinical language and a coherent structure. However, it ignores critical information such as test results, comorbidities, uncertainty, and alternative diagnoses. This mirrors concerns about large language models in healthcare, where outputs may sound authoritative while lacking clinical grounding or evidence. Without safeguards and oversight, such systems risk misleading clinicians or patients.
